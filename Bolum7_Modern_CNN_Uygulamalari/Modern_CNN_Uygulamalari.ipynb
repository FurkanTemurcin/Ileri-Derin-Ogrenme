```python
import tensorflow as tf
import matplotlib.pyplot as plt

# 1. Veri Setini Yükleme ve tf.data Pipeline'ı Kurma
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()

def preprocess(image, label):
    image = tf.expand_dims(image, -1) # Kanal boyutu ekle
    image = tf.image.resize(image, [224, 224]) # AlexNet için boyutu büyüt
    image = tf.cast(image, tf.float32) / 255.0 # Normalize et
    return image, label

batch_size = 128
# Veriyi batch'ler halinde arka planda hazırlaması için AUTOTUNE kullanıyoruz
train_ds = tf.data.Dataset.from_tensor_slices((x_train, y_train)).map(preprocess, num_parallel_calls=tf.data.AUTOTUNE).batch(batch_size).prefetch(tf.data.AUTOTUNE)
test_ds = tf.data.Dataset.from_tensor_slices((x_test, y_test)).map(preprocess, num_parallel_calls=tf.data.AUTOTUNE).batch(batch_size).prefetch(tf.data.AUTOTUNE)

# 2. AlexNet Mimarisi
net = tf.keras.models.Sequential([
    tf.keras.layers.Conv2D(filters=96, kernel_size=11, strides=4, activation='relu', input_shape=(224, 224, 1)),
    tf.keras.layers.MaxPool2D(pool_size=3, strides=2),
    tf.keras.layers.Conv2D(filters=256, kernel_size=5, padding='same', activation='relu'),
    tf.keras.layers.MaxPool2D(pool_size=3, strides=2),
    tf.keras.layers.Conv2D(filters=384, kernel_size=3, padding='same', activation='relu'),
    tf.keras.layers.Conv2D(filters=384, kernel_size=3, padding='same', activation='relu'),
    tf.keras.layers.Conv2D(filters=256, kernel_size=3, padding='same', activation='relu'),
    tf.keras.layers.MaxPool2D(pool_size=3, strides=2),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(4096, activation='relu'),
    tf.keras.layers.Dropout(0.5),
    tf.keras.layers.Dense(4096, activation='relu'),
    tf.keras.layers.Dropout(0.5),
    tf.keras.layers.Dense(10, activation='softmax')
])

# 3. Model Özeti ve Katman Çıktı Şekilleri
print("--- AlexNet Model Özeti ---")
net.summary()

print("\n--- Katman Çıktı Şekilleri ---")
X = tf.random.uniform((1, 224, 224, 1))
for layer in net.layers:
    X = layer(X)
    print(layer.__class__.__name__, 'çıktı şekli:\t', X.shape)

# 4. Derleme ve Eğitim
net.compile(optimizer=tf.keras.optimizers.SGD(learning_rate=0.01),
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

print("\nEğitim Başlıyor (10 Epoch)...")
history = net.fit(train_ds, epochs=10, validation_data=test_ds)

# 5. Görselleştirme
plt.figure(figsize=(10, 6))
plt.plot(history.history['accuracy'], label='Eğitim Doğruluğu', marker='o')
plt.plot(history.history['val_accuracy'], label='Test Doğruluğu', marker='o')
plt.title('AlexNet - Fashion MNIST Eğitim Eğrisi')
plt.ylabel('Doğruluk')
plt.xlabel('Dönem (Epoch)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import tensorflow as tf
import matplotlib.pyplot as plt

# 1. Veri Hazırlığı
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()

def preprocess(image, label):
    image = tf.expand_dims(image, -1)
    image = tf.image.resize(image, [224, 224])
    image = tf.cast(image, tf.float32) / 255.0
    return image, label

batch_size = 128
train_ds = tf.data.Dataset.from_tensor_slices((x_train, y_train)).map(preprocess, num_parallel_calls=tf.data.AUTOTUNE).batch(batch_size).prefetch(tf.data.AUTOTUNE)
test_ds = tf.data.Dataset.from_tensor_slices((x_test, y_test)).map(preprocess, num_parallel_calls=tf.data.AUTOTUNE).batch(batch_size).prefetch(tf.data.AUTOTUNE)

# 2. VGG Bloğu Tanımlama Fonksiyonu
def vgg_block(num_convs, num_channels):
    blk = tf.keras.models.Sequential()
    for _ in range(num_convs):
        blk.add(tf.keras.layers.Conv2D(num_channels, kernel_size=3, padding='same', activation='relu'))
    blk.add(tf.keras.layers.MaxPool2D(pool_size=2, strides=2))
    return blk

# 3. VGG-11 Yapılandırması ve Küçültülmesi
conv_arch = ((1, 64), (1, 128), (2, 256), (2, 512), (2, 512))
ratio = 4
small_conv_arch = [(pair[0], pair[1] // ratio) for pair in conv_arch]

# 4. Modeli İnşa Etme
def vgg(conv_arch_config):
    net = tf.keras.models.Sequential()
    # Evrişimli blokları döngü ile ekleme
    for (num_convs, num_channels) in conv_arch_config:
        net.add(vgg_block(num_convs, num_channels))
    # Tam bağlı (Dense) kısmı ekleme
    net.add(tf.keras.models.Sequential([
        tf.keras.layers.Flatten(),
        tf.keras.layers.Dense(4096, activation='relu'),
        tf.keras.layers.Dropout(0.5),
        tf.keras.layers.Dense(4096, activation='relu'),
        tf.keras.layers.Dropout(0.5),
        tf.keras.layers.Dense(10, activation='softmax')
    ]))
    return net

model_vgg = vgg(small_conv_arch)

# 5. Model Özeti ve Katman Çıktı Şekilleri
model_vgg.build(input_shape=(None, 224, 224, 1))

print("--- VGG-11 (Küçültülmüş) Model Özeti ---")
model_vgg.summary()

print("\n--- Katman Çıktı Şekilleri ---")
X = tf.random.uniform((1, 224, 224, 1))
for blk in model_vgg.layers:
    X = blk(X)
    print(blk.__class__.__name__,'çıktı şekli:\t', X.shape)

# 6. Derleme ve Eğitim
model_vgg.compile(optimizer=tf.keras.optimizers.SGD(learning_rate=0.05),
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])

print("\nVGG-11 Eğitimi Başlıyor (10 Epoch)...")
history_vgg = model_vgg.fit(train_ds, epochs=10, validation_data=test_ds)

# 7. Görselleştirme
plt.figure(figsize=(10, 6))
plt.plot(history_vgg.history['accuracy'], label='VGG Eğitim Doğruluğu', marker='o')
plt.plot(history_vgg.history['val_accuracy'], label='VGG Test Doğruluğu', marker='o')
plt.title('VGG-11 (Küçültülmüş) - Fashion MNIST Eğitim Eğrisi')
plt.ylabel('Doğruluk')
plt.xlabel('Dönem (Epoch)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import tensorflow as tf
import matplotlib.pyplot as plt

# 1. Veri Hazırlığı
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()

def preprocess(image, label):
    image = tf.expand_dims(image, -1)
    image = tf.image.resize(image, [224, 224])
    image = tf.cast(image, tf.float32) / 255.0
    return image, label

batch_size = 128
train_ds = tf.data.Dataset.from_tensor_slices((x_train, y_train)).map(preprocess, num_parallel_calls=tf.data.AUTOTUNE).batch(batch_size).prefetch(tf.data.AUTOTUNE)
test_ds = tf.data.Dataset.from_tensor_slices((x_test, y_test)).map(preprocess, num_parallel_calls=tf.data.AUTOTUNE).batch(batch_size).prefetch(tf.data.AUTOTUNE)

# 2. NiN Bloğu Tanımlaması
def nin_block(num_channels, kernel_size, strides, padding):
    return tf.keras.models.Sequential([
        tf.keras.layers.Conv2D(num_channels, kernel_size, strides=strides, padding=padding, activation='relu'),
        tf.keras.layers.Conv2D(num_channels, kernel_size=1, activation='relu'), # Piksel başına MLP
        tf.keras.layers.Conv2D(num_channels, kernel_size=1, activation='relu')  # Piksel başına MLP
    ])

# 3. NiN Mimarisinin İnşası
def net_nin():
    return tf.keras.models.Sequential([
        nin_block(96, kernel_size=11, strides=4, padding='valid'),
        tf.keras.layers.MaxPool2D(pool_size=3, strides=2),
        nin_block(256, kernel_size=5, strides=1, padding='same'),
        tf.keras.layers.MaxPool2D(pool_size=3, strides=2),
        nin_block(384, kernel_size=3, strides=1, padding='same'),
        tf.keras.layers.MaxPool2D(pool_size=3, strides=2),
        tf.keras.layers.Dropout(0.5),

        # Son katmanda etiket sınıfı sayısı (Fashion-MNIST için 10) kadar kanal oluşturulur.
        nin_block(10, kernel_size=3, strides=1, padding='same'),

        # Hayati adım: Global Average Pooling
        tf.keras.layers.GlobalAveragePooling2D(),

        # Sonucu Keras formatına uygun şekilde (Batch_size, 10) vektörüne dönüştür
        tf.keras.layers.Flatten(),
        tf.keras.layers.Activation('softmax') # Sınıflandırma çıktısı için
    ])

model_nin = net_nin()

# 4. Model Özeti ve Katman Çıktı Şekilleri
model_nin.build(input_shape=(None, 224, 224, 1))

print("--- NiN (Ağ İçinde Ağ) Model Özeti ---")
model_nin.summary()

print("\n--- NiN Katman Çıktı Şekilleri ---")
X = tf.random.uniform((1, 224, 224, 1))
for layer in model_nin.layers:
    X = layer(X)
    print(layer.__class__.__name__,'çıktı şekli:\t', X.shape)

# 5. Derleme ve Eğitim (Kitaptaki gibi lr=0.1 kullanıyoruz)
model_nin.compile(optimizer=tf.keras.optimizers.SGD(learning_rate=0.1),
                  loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])

print("\nNiN Eğitimi Başlıyor (10 Epoch)...")
history_nin = model_nin.fit(train_ds, epochs=10, validation_data=test_ds)

# 6. Görselleştirme
plt.figure(figsize=(10, 6))
plt.plot(history_nin.history['accuracy'], label='NiN Eğitim Doğruluğu', marker='o')
plt.plot(history_nin.history['val_accuracy'], label='NiN Test Doğruluğu', marker='o')
plt.title('NiN (Ağ İçinde Ağ) - Fashion MNIST Eğitim Eğrisi')
plt.ylabel('Doğruluk')
plt.xlabel('Dönem (Epoch)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import tensorflow as tf
import matplotlib.pyplot as plt

# 1. Veri Hazırlığı (GoogLeNet için boyutu 96x96'ya düşürüyoruz)
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()

def preprocess(image, label):
    image = tf.expand_dims(image, -1)
    image = tf.image.resize(image, [96, 96]) # Eğitim hızlansın diye 96x96
    image = tf.cast(image, tf.float32) / 255.0
    return image, label

batch_size = 128
train_ds = tf.data.Dataset.from_tensor_slices((x_train, y_train)).map(preprocess, num_parallel_calls=tf.data.AUTOTUNE).batch(batch_size).prefetch(tf.data.AUTOTUNE)
test_ds = tf.data.Dataset.from_tensor_slices((x_test, y_test)).map(preprocess, num_parallel_calls=tf.data.AUTOTUNE).batch(batch_size).prefetch(tf.data.AUTOTUNE)

# 2. Inception Bloğu Tanımlaması
class Inception(tf.keras.Model):
    def __init__(self, c1, c2, c3, c4):
        super().__init__()
        # Yol 1
        self.p1_1 = tf.keras.layers.Conv2D(c1, 1, activation='relu')
        # Yol 2 (1x1 Bottleneck + 3x3)
        self.p2_1 = tf.keras.layers.Conv2D(c2[0], 1, activation='relu')
        self.p2_2 = tf.keras.layers.Conv2D(c2[1], 3, padding='same', activation='relu')
        # Yol 3 (1x1 Bottleneck + 5x5)
        self.p3_1 = tf.keras.layers.Conv2D(c3[0], 1, activation='relu')
        self.p3_2 = tf.keras.layers.Conv2D(c3[1], 5, padding='same', activation='relu')
        # Yol 4
        self.p4_1 = tf.keras.layers.MaxPool2D(3, 1, padding='same')
        self.p4_2 = tf.keras.layers.Conv2D(c4, 1, activation='relu')

    def call(self, x):
        p1 = self.p1_1(x)
        p2 = self.p2_2(self.p2_1(x))
        p3 = self.p3_2(self.p3_1(x))
        p4 = self.p4_2(self.p4_1(x))
        # 4 Paralel yolun Kanal Ekseni üzerinden BİRLEŞTİRİLMESİ (Concatenate)
        return tf.keras.layers.Concatenate(axis=-1)([p1, p2, p3, p4])

# 3. GoogLeNet Modüllerinin İnşası (Eksiksiz 5 Modül)
def b1():
    return tf.keras.Sequential([
        tf.keras.layers.Conv2D(64, 7, strides=2, padding='same', activation='relu'),
        tf.keras.layers.MaxPool2D(pool_size=3, strides=2, padding='same')])

def b2():
    return tf.keras.Sequential([
        tf.keras.layers.Conv2D(64, 1, activation='relu'),
        tf.keras.layers.Conv2D(192, 3, padding='same', activation='relu'),
        tf.keras.layers.MaxPool2D(pool_size=3, strides=2, padding='same')])

def b3():
    return tf.keras.Sequential([
        Inception(64, (96, 128), (16, 32), 32),
        Inception(128, (128, 192), (32, 96), 64),
        tf.keras.layers.MaxPool2D(pool_size=3, strides=2, padding='same')])

def b4():
    return tf.keras.Sequential([
        Inception(192, (96, 208), (16, 48), 64),
        Inception(160, (112, 224), (24, 64), 64),
        Inception(128, (128, 256), (24, 64), 64),
        Inception(112, (144, 288), (32, 64), 64),
        Inception(256, (160, 320), (32, 128), 128),
        tf.keras.layers.MaxPool2D(pool_size=3, strides=2, padding='same')])

def b5():
    return tf.keras.Sequential([
        Inception(256, (160, 320), (32, 128), 128),
        Inception(384, (192, 384), (48, 128), 128),
        tf.keras.layers.GlobalAvgPool2D(), # NiN felsefesi aynen devam
        tf.keras.layers.Flatten()
    ])

def net_googlenet():
    return tf.keras.Sequential([b1(), b2(), b3(), b4(), b5(), tf.keras.layers.Dense(10, activation='softmax')])

model_googlenet = net_googlenet()

# 4. Model Özeti ve Çıktı Şekilleri
model_googlenet.build(input_shape=(None, 96, 96, 1))
print("--- GoogLeNet (Inception) Model Özeti ---")
model_googlenet.summary()

print("\n--- GoogLeNet Katman Çıktı Şekilleri ---")
X = tf.random.uniform((1, 96, 96, 1))
for layer in model_googlenet.layers:
    X = layer(X)
    print(layer.__class__.__name__, 'çıktı şekli:\t', X.shape)

# 5. Derleme ve Eğitim (Kitaptaki gibi lr=0.1)
model_googlenet.compile(optimizer=tf.keras.optimizers.SGD(learning_rate=0.1),
                        loss='sparse_categorical_crossentropy',
                        metrics=['accuracy'])

print("\nGoogLeNet Eğitimi Başlıyor (10 Epoch)...")
history_googlenet = model_googlenet.fit(train_ds, epochs=10, validation_data=test_ds)

# 6. Görselleştirme
plt.figure(figsize=(10, 6))
plt.plot(history_googlenet.history['accuracy'], label='GoogLeNet Eğitim Doğruluğu', marker='o')
plt.plot(history_googlenet.history['val_accuracy'], label='GoogLeNet Test Doğruluğu', marker='o')
plt.title('GoogLeNet (Inception) - Fashion MNIST Eğitim Eğrisi')
plt.ylabel('Doğruluk')
plt.xlabel('Dönem (Epoch)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np

# 1. Veri Hazırlığı (LeNet orijinal 28x28 boyutunu kabul eder, boyutlandırmaya gerek yok)
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()

def preprocess(image, label):
    image = tf.expand_dims(image, -1)
    image = tf.cast(image, tf.float32) / 255.0
    return image, label

batch_size = 256
train_ds = tf.data.Dataset.from_tensor_slices((x_train, y_train)).map(preprocess, num_parallel_calls=tf.data.AUTOTUNE).batch(batch_size).prefetch(tf.data.AUTOTUNE)
test_ds = tf.data.Dataset.from_tensor_slices((x_test, y_test)).map(preprocess, num_parallel_calls=tf.data.AUTOTUNE).batch(batch_size).prefetch(tf.data.AUTOTUNE)

# 2. Toplu Normalleştirmeli LeNet Mimarisi (Kısa Uygulama - Keras API)
def net_batchnorm():
    return tf.keras.models.Sequential([
        tf.keras.layers.Conv2D(filters=6, kernel_size=5, input_shape=(28, 28, 1)),
        tf.keras.layers.BatchNormalization(), # Evrişimden SONRA, Aktivasyondan ÖNCE
        tf.keras.layers.Activation('sigmoid'),
        tf.keras.layers.AvgPool2D(pool_size=2, strides=2),
        tf.keras.layers.Conv2D(filters=16, kernel_size=5),
        tf.keras.layers.BatchNormalization(),
        tf.keras.layers.Activation('sigmoid'),
        tf.keras.layers.AvgPool2D(pool_size=2, strides=2),
        tf.keras.layers.Flatten(),
        tf.keras.layers.Dense(120),
        tf.keras.layers.BatchNormalization(),
        tf.keras.layers.Activation('sigmoid'),
        tf.keras.layers.Dense(84),
        tf.keras.layers.BatchNormalization(),
        tf.keras.layers.Activation('sigmoid'),
        tf.keras.layers.Dense(10, activation='softmax')
    ])

model_bn = net_batchnorm()

# 3. İntihar Görevi: YÜKSEK ÖĞRENME ORANI İLE EĞİTİM (lr=1.0)
model_bn.compile(optimizer=tf.keras.optimizers.SGD(learning_rate=1.0),
                 loss='sparse_categorical_crossentropy',
                 metrics=['accuracy'])

print("--- Toplu Normalleştirmeli LeNet Modeli (lr=1.0) Eğitimi Başlıyor ---")
history_bn = model_bn.fit(train_ds, epochs=10, validation_data=test_ds)

# 4. Görselleştirme (Slayta Eklenecek Grafik)
plt.figure(figsize=(10, 6))
plt.plot(history_bn.history['accuracy'], label='Eğitim Doğruluğu (Batch Norm)', marker='o')
plt.plot(history_bn.history['val_accuracy'], label='Test Doğruluğu (Batch Norm)', marker='o')
plt.title('Toplu Normalleştirmeli LeNet Eğitimi (Absürt Öğrenme Oranı: 1.0)')
plt.ylabel('Doğruluk')
plt.xlabel('Dönem (Epoch)')
plt.legend()
plt.grid(True)
plt.show()

# Modeldeki ilk BatchNormalization katmanı index 1'dedir (index 0 Conv2D)
gamma_weights = model_bn.layers[1].gamma.numpy()
beta_weights = model_bn.layers[1].beta.numpy()

print("\n===========================================================")
print("===========================================================")
print("1. Evrişim Katmanından Sonraki İlk Batch Norm'un Öğrendiği Değerler:\n")
print(f"Öğrenilmiş Gamma (Ölçek/Scale) Değerleri:\n {np.round(gamma_weights, 4)}")
print(f"\nÖğrenilmiş Beta (Kaydırma/Shift) Değerleri:\n {np.round(beta_weights, 4)}")
print("===========================================================")
```
